```python
#TASK-1 Q3
try:
    hill_number = int(input("ENTER ANY NUMBER TO CHECK WHETHER IT IS A HILL NUMBER OR NOT:"))
    k = str(hill_number)
    is_hill = True
    for i in range(len(k)//2):
        j = -1-i
        if not k[j]<k[j-1] and k[i]<k[i+1] and k[i]==k[j]:
         is_hill = False
         break
    if (len(k)>=3 and is_hill and len(k)%2 == 0 and k[len(k)//2] == k[(len(k)//2)-1]) or (len(k)>=3 and is_hill and len(k)%2 != 0 and k[len(k)//2]>k[len(k)//2-1] and k[len(k)//2]>k[-(len(k)//2)]):
     print("The entered number is a Hill Number.")
    else: 
     print("The entered number is not a Hill Number.")
except ValueError:
    print("Only Integers should be written.")
```

```python
#TASK-1 Q4
word = input("Enter any word to get it sorted in alphabetical order").upper()
word_list=list(word)
word_sorted=[]
while word_list:
    min_word=min(word_list)
    word_sorted.append(min_word)
    word_list.remove(min_word)
print("".join(word_sorted))
```

```python
#TASK-1 Q5
def fibonacci(n):
  if n <= 0:
    return 0
  elif n == 1:
    return 1
  return fibonacci(n - 1) + fibonacci(n - 2)
```

```python
fibonacci(7)
```
