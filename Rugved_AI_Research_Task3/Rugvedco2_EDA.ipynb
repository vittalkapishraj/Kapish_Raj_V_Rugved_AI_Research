```python
df_cleaning=df.copy()
```

```python
df_cleaning
```

```python
for col in df_cleaning.select_dtypes(include=[np.number]).columns:
         lower = df_cleaning[col].quantile(0.1)
         upper = df_cleaning[col].quantile(0.9)
         df_cleaning[col] = np.clip(df_cleaning[col], lower, upper)
```

```python
df_cleaning.describe()
```

```python
df_cleaning.skew(numeric_only=True)
```

```python
df_cleaning["Engine Size(L)"]=np.clip(df_cleaning["Engine Size(L)"],df_cleaning["Engine Size(L)"].quantile(0.2),df_cleaning['Engine Size(L)'].quantile(0.8))
```
