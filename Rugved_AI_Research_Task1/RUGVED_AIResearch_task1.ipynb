```python
#TASK-1 Q5
def fibonacci(n):
  if n <= 0:
    return 0
  elif n == 1:
    return 1
  return fibonacci(n - 1) + fibonacci(n - 2)
```

```python
fibonacci(7)
```

```python
#TASK-1 Q8
string = input("Enter the string for which you want to check it has equal parts or not and also are those parts same.")
n = int(input("Enter number with which you want to check it has equal parts or not and were those parts same."))
k=string[:n]
m=[]
if len(string)%n == 0:
    for i in range(0,len(string)//n):
        x = string[(i)*n:(i+1)*n]
        if x == k:
            m.append(x)
        else:
            print("Appropriate Error")
            m=[]
            break
else:
    print("Appropriate Error")
print(",".join(m))
            
            
```

```python
#Task-1 Q9
text = input("Enter a word or a sentence to encrypt")
type_of_cryption = input("Enter you want to encrypt or decrypt").upper()
shift_number = int(input("Enter number by which you want to shift."))
if type_of_cryption == "ENCRYPT":
    shift = shift_number
elif type_of_cryption == "DECRYPT":
    shift = -(shift_number)
else:
    print("Enter only the mode of encryption as encrypt or decrypt")
final = ""
for char in text:
        if char.isalpha():
            if char.isupper():
                start = ord("A")
            else:
                start = ord('a')
            new_char = chr((ord(char) - start + shift) % 26 + start)
            final += new_char
        else:
            final += char
print(final)
    

    
```

```python
#TASK-1 Q10
number = input("Enter your credit card number:\n ")
k = len(number)
sum_digits = 0
main = number[1:k+1:2]
for char in main:
   i = int(char)*2
   if int(i)<=9:
       sum_digits+=int(i)
   else:
       for j in str(i):
           sum_digits+=int(j)
casual = number[0:k+1:2]
for dig in casual:
    sum_digits+=int(dig)
if sum_digits%10==0:
    print("Valid card number")
else:
    print("Invalid card number") 
```

```python
#TASK-1 Q11
text = input("Text: ")
letters = 0
words = 0
sentences = 0
word = False
    
for char in text:
    if char.isalpha():
        letters += 1
    if char in ['.', '!', '?']:
        sentences += 1
    if not char.isspace():
        if not word:
            words += 1
            word = True
    else:
        word = False
if words == 0:
    print("Before Grade 1")
L = (letters / words) * 100
S = (sentences / words) * 100
grade_point = 0.0588 * L - 0.296 * S - 15.8
grade = round(grade_point )
if grade >= 16:
    print("Grade 16+")
elif grade < 1:
    print("Before Grade 1")
else:
    print(f"Grade {grade}")
```
