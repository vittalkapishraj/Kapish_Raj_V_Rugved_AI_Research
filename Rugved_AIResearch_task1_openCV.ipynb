```python
 import cv2
```

```python
img = cv2.imread("sample.img.jpeg")
```

```python
import matplotlib.pyplot as plt
```

```python
print(f"Height of image:{img.shape[0]} pixels \nWidth of image:{img.shape[1]} pixels \nNumber of channels:{img.shape[2]}")
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) 
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis('off')
plt.show()
```

```python
cv2.imwrite('saved_sample_image.jpg', img)
print("Image saved as 'saved_sample_image.jpg'")
```

```python
resized = cv2.resize(img, (300, 300))
cv2.imwrite('resized_sample.jpg', resized)
if True:
    print("resized_sample.jpg saved")
else:
    print("An error occured")
```

```python
cropped = img[50:200, 50:200]
cv2.imwrite('cropped_sample.jpg', cropped)
if True:
    print("cropped_sample.jpg saved")
else:
    print("An error occured")
```

```python
flipped = cv2.flip(img, 1) 
cv2.imwrite('flipped_sample.jpg', flipped)
if True:
    print("flipped_sample.jpg saved")
else:
    print("An error occured")
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
plt.title("Resized")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB))
plt.title("Cropped")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(flipped, cv2.COLOR_BGR2RGB))
plt.title("Flipped")
plt.axis('off')

plt.show()
```

```python
grayed_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2.imwrite('gray_sample.jpg', grayed_img)
```

```python
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
cv2.imwrite('hsv_sample.jpg', hsv)
```

```python
edges = cv2.Canny(grayed_img, 100, 200)
cv2.imwrite('edges_sample.jpg', edges)
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 4, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(grayed_img, cmap='gray')
plt.title("Grayscale")
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(hsv)
plt.title("HSV")
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(edges, cmap='gray')
plt.title("Edges (Canny)")
plt.axis('off')

plt.show()
```
